```python
high_potential = [
    'potential',
    'long term',
    'in progress positive',
    'open',
    'converted'
]
low_potential = [
    'junk lead',
    'not responding',
    'just enquiry',
    'in progress negative',
    'lost'
]

df['lead_category'] = np.nan


df.loc[df['status_clean'].isin(high_potential),'lead_category'] = 'high potential'
df.loc[df['status_clean'].isin(low_potential),'lead_category']  = 'low potential'
```

********Data Visualization********

```python
plt.figure(figsize = (10,6))

sns.countplot(
    data = df,
    x = 'status_clean'
)
plt.title('Lead Status Distribution')
plt.xlabel('Status')
plt.ylabel('Number of lead')
plt.xticks(rotation =90)
plt.tight_layout()
```

```python
top_sources = df['Source'].value_counts().head(10).index
plt.figure(figsize=(10,6))

sns.countplot(
    data = df[df['Source'].isin(top_sources)],
    y = 'Source',
    order = top_sources
)
plt.title('Top 10 Sources')
plt.xlabel('Number of leads')
plt.ylabel('Lead Sourses')
plt.xticks(rotation =90)
plt.tight_layout()
plt.show()
```

```python
top_location = df['Location'].value_counts().head(10).index

plt.figure(figsize = (10,6))

sns.countplot(
    data = df[df['Location'].isin(top_location)],
    y = 'Location',
    order = top_location
)

plt.title('Top 10 Location')
plt.xlabel('Number of leads')
plt.ylabel('Locations')
plt.xticks(rotation =90)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data = df,
    x = 'Delivery_Mode'
)

plt.title('Top Delivery Mode')
plt.xlabel('Delivery Mode')
plt.ylabel('Number of lead')
plt.xticks(rotation =90)
plt.tight_layout()
```

```python
source_status = pd.crosstab(df['Delivery_Mode'],df['status_clean'])

plt.figure(figsize=(14, 10))

sns.heatmap(
    source_status,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title("Lead Delivery Mode Distubution by Status")
plt.xlabel('Status')
plt.ylabel('Delivery Mode ')
plt.tight_layout()
plt.show()
```

```python
source_status = pd.crosstab(df['Sales_Agent'],df['status_clean'])

plt.figure(figsize=(14, 10))

sns.heatmap(
    source_status,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title("Lead Sales Distibution by Status")
plt.xlabel('Status')
plt.ylabel('Sales ')
plt.tight_layout()
plt.show()
```

```python
source_status = pd.crosstab(df['Location'],df['status_clean'])

plt.figure(figsize=(14, 10))

sns.heatmap(
    source_status,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title("Lead Location Distribution by  Status")
plt.xlabel('Status')
plt.ylabel('Location')
plt.tight_layout()
plt.show()
```

```python

source_status = pd.crosstab(df['Source'],df['status_clean'])

plt.figure(figsize=(14, 10))

sns.heatmap(
    source_status,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title("Lead Source Distrubution by Status")
plt.xlabel('Status')
plt.ylabel('Source')
plt.tight_layout()
plt.show()
```
