```python

source_status = pd.crosstab(df['Source'],df['Status'])

plt.figure(figsize=(14, 10))

sns.heatmap(
    source_status,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title("Source vs Status")
plt.xlabel('Status')
plt.ylabel('Source')
plt.tight_layout()
plt.show()
```
